```python

import pandas as pd
from pathlib import Path
```

```python
data_path = Path("../data/raw/DataCoSupplyChainDataset.csv")
df = pd.read_csv(data_path, encoding="latin-1")
```

```python
print(df.shape)
```

```python
print(df.head())
print(df.columns.tolist())
```

```python
data_dictionary = pd.DataFrame({
    "column_name": df.columns,
    "Observed data_type": [str(df[col].dtype) for col in df.columns],
    "missing_values": [df[col].isna().sum() for col in df.columns],
    "missing_percentage": [
        round(df[col].isna().mean() * 100, 2)
        for col in df.columns
    ],
    "unique_values": [
        df[col].nunique(dropna=True) for col in df.columns
    ],
    "description": ["" for col in df.columns],
    "validation_rules": ["" for col in df.columns]
})

print(data_dictionary)

print("Data dictionary created successfully!")
```
